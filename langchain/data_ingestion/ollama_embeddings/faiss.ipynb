```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

loader=TextLoader("speech.txt")
documents=loader.load()
```

```python
chunk=CharacterTextSplitter(chunk_size=500,chunk_overlap=30)
docs=chunk.split_documents(documents)
```

```python
docs
```

```python
embdeddings=OllamaEmbeddings(model="nomic-embed-text")
```

```python
db=FAISS.from_documents(docs,embdeddings)
```

```python
db
```

```python
query="How data is collected and used in the speech?"
query_result=db.similarity_search(query)
```

```python
query_result
```

```python
query_result[0].page_content
```

```python
### vector store as a retriever 
##Retriever = librarian jo aapke question ke basis par relevant books/chunks nikalta hai
retriever=db.as_retriever()
docs=retriever.invoke(query)
```

```python
docs[0].page_content
```

```python
## similarity search with score(lower is best)
docs_and_score=db.similarity_search_with_score(query)
```

```python
docs_and_score
```

```python
embeddings_vector=embdeddings.embed_query(query)
```

```python
docs_score=db.similarity_search_by_vector(embeddings_vector)
docs_score
```

```python
## saving and loading
db.save_local("faiss_index")
```

```python
new_df=FAISS.load_local("faiss_index",embdeddings,allow_dangerous_deserialization=True)
```

```python
#### chroma db

from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
data=loader.load()
data
```

```python
chunk=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=30)
docs=chunk.split_documents(data)
```

```python
docs
```

```python
embeddings=OllamaEmbeddings(model="nomic-embed-text")
db=Chroma.from_documents(documents=docs,embedding=embeddings)
```

```python
db
```

```python
query="What is the main topic of the speech?"
db.similarity_search_with_score(query)
```

```python
docs[0].page_content
```

```python
vectordb=Chroma.from_documents(documents=docs,embedding=embeddings,persist_directory="./chroma_db")
```

```python
db2=Chroma(persist_directory="./chroma_db",embedding_function=embeddings)
docs=db2.similarity_search(query)
print(docs[0].page_content)
```

```python
retriever=vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
