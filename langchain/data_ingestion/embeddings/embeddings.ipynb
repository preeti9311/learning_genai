```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
vector = embeddings.embed_query("This is a sample text to be embedded.")
print(len(vector))
```

```python
print(embeddings.model)
```

```python
vector1024 = embeddings.embed_query("This is a sample text to be embedded.",dimensions=1024)
print(len(vector1024))
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("speech.txt")
docs=loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=50)
final_docs=text_splitter.split_documents(docs)
final_docs
```

```python
## string to the db
## using chroma db
from langchain_community.vectorstores import Chroma
db=Chroma.from_documents(final_docs,embeddings)
db
```

```python
query="The quality of any data-driven system depends heavily on the quality of its data ingestion process."
retrieved_results=db.similarity_search(query)
print(retrived_results)
```
